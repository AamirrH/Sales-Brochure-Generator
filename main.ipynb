```python
from scraper import fetch_website_links
from openai import OpenAI
from dotenv import load_dotenv
import os
import json
```

```python
## Initialise Environment and Gemini
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
load_dotenv(override=True)
google_api_key = os.getenv("GEMINI_API_KEY")
## Load Gemini
gemini = OpenAI(api_key=google_api_key,base_url=GEMINI_BASE_URL)

```

```python
## Provide the Primary company website, you want to create a brochure for
link = input("Provide the Link here :- ")
## Fetches all the subsequent links from a primary website page
links = fetch_website_links(link)
print(links)
```

```python

## Now we have to give all these links to an LLM model and make it figure it which links are actually relevant to the primary website.
link_system_prompt = """
You are provided with a list of links and you will also be provided with a primary website link
as well. Your task is to return a list of links that are relevant to the primary company website link
and remove all the unecessary ones.
You should include links like About Us/ Social Media Handles and other relevant links.
Your job is to figure out what links to return so that
a good Sales Brochure about the company can be made 
using that information and return the list in a JSON format like this.

{
  "links" : [
  {"type": "about page", "url": *provide link here* }
  {"type": "media page", "url": *provide link here* }
  ]

}
"""
```

```python
def get_relevant_links(links,system_prompt):
    response = gemini.chat.completions.create(messages=[{
        "role" : "system",
        "content" : system_prompt
    },
    {
        "role" : "user",
        "content" : links
    }],model="gemini-flash-latest")
    print(response.choices[0].message.content,)
```

```python
get_relevant_links(links=links,system_prompt=link_system_prompt)
```
