```python
from scraper import fetch_website_links
from openai import OpenAI
from dotenv import load_dotenv
import os
import json
from IPython.display import Markdown, display
```

```python
## Initialise Environment and Gemini
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
load_dotenv(override=True)
google_api_key = os.getenv("GEMINI_API_KEY")
## Load Gemini
gemini = OpenAI(api_key=google_api_key,base_url=GEMINI_BASE_URL)

```

```python
## Provide the Primary company website, you want to create a brochure for
link = input("Provide the Link here :- ")
## Fetches all the subsequent links from a primary website page
links = fetch_website_links(link)
print(links)
```

```python

## Now we have to give all these links to an LLM model and make it figure it which links are actually relevant to the primary website.
link_system_prompt = """
You are provided with a list of links and you will also be provided with a primary website link
as well. Your task is to return a list of links that are relevant to the primary company website link
and remove all the unecessary ones.
You should include links like About Us/ Social Media Handles and other relevant links.
Your job is to figure out what links to return so that
a good Sales Brochure about the company can be made 
using that information and return the list in a JSON format like this.

{
  "links" : [
  {"type": "about page", "url": *provide link here* }
  {"type": "media page", "url": *provide link here* }
  ]

}
"""
```

```python
def get_relevant_links(primary_url, links, system_prompt):
    user_prompt = {
        "primary_website": primary_url,
        "links": links
    }

    response = gemini.chat.completions.create(
        model="gemini-flash-latest",
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": json.dumps(user_prompt, indent=2)
            }
        ],
        response_format={"type": "json_object"}
    )

    return json.loads(response.choices[0].message.content)
```

```python
relevant_links = get_relevant_links(primary_url=link,links=links,system_prompt=link_system_prompt)
```

```python
## Now using the get relevant links function we will create a brochure.
brochure_prompt = """
You are an Expert Sales Brochure Creator, when provided with a list of all the relevant links,
you will extract information from the links and create a unique and high impact sales brochure
about the primary product/website. you will receive the relevant links in the form of JSON
Your task is to go through the links and create the Sales Brochure
Keep in mind, the product could be anything, even someones personal website, in that case
you have to create a brochure for the value the person provides not on the person itself
because brochures are created for products.
You can create the result in Markdown, so as to look good.
"""
```

```python
def create_sales_brochure(brochure_prompt, relevant_links):
    markdown_response = gemini.chat.completions.create(
        model="gemini-flash-latest",
        messages=[
            {
                "role": "system",
                "content": brochure_prompt
            },
            {
                "role": "user",
                "content": json.dumps(relevant_links, indent=2)
            }
        ]
    )

    return markdown_response.choices[0].message.content
```

```python
sales_brochure = create_sales_brochure(brochure_prompt=brochure_prompt,relevant_links=relevant_links)
display(Markdown(sales_brochure))
```
